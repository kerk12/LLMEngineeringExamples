```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="gpt-oss:20b")

messages = []
messages
```

```python
messages.append({
    "role": "user",
    "content": "What is the name of the protagonist of Hunger Games?",
})

response = model.invoke(input=messages)
response.pretty_print()
```

```python
messages.append({
    "role": "assistant",
    "content": response.content
})
messages
```

```python
messages.append({
    "role": "user",
    "content": "Has this actor starred in any other noteworthy movies?"
})

response = model.invoke(input=messages)
messages.append({
    "role": "assistant", "content": response.content
})

messages
```
