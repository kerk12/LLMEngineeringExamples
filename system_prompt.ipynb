## Παράδειγμα χρήσης System Prompt

````python
from langchain_ollama import ChatOllama

model = ChatOllama(model="gpt-oss:20b")

# Ορίζουμε το system prompt:
messages = [
    {
        "role": "system",
        "content": """
        Είσαι βοηθός γιατρού. Δουλειά σου είναι να προγραμματίζεις ραντεβού και να βοηθάς τους ασθενείς
        με απορίες που μπορεί να έχουν σχετικά με τις υπηρεσίες του ιατρείου.

        Θα απαντάς πάντα με επίσημο ύφος.

        Απαγορεύεται να δίνεις ιατρικές συμβουλές.
        Το ιατρείο δεν πουλά το οποιοδήποτε προϊόν.

        Αν δεν γνωρίζεις κάτι, μην αυτοσχεδιάσεις και πες οτι δεν το γνωρίζεις.
        """
    }
]
````

````python
# Καλούμε το LLM, δίνοντάς του το system prompt και το μήνυμά μας.
response = model.invoke(input=[
    *messages,
    {
        "role": "user",
        "content": "Τι υπηρεσίες διαθέτει το ιατρείο;"
    }
])
# Η απάντηση του LLM
response.pretty_print()
````

````python
# Κακή απορία από τον χρήστη (ζητάει ιατρική συμβουλή)
response = model.invoke(input=[
    *messages,
    {
        "role": "user",
        "content": "Πονάει το πόδι μου. Σε τι μπορεί να οφείλεται; Μπορείτε να με συμβουλέψετε;"
    }
])
# Η απάντηση του LLM
response.pretty_print()
````

````python
# Κακή απορία από τον χρήστη (ρωτάει άσχετη ερώτηση)
response = model.invoke(input=[
    *messages,
    {
        "role": "user",
        "content": "Διαθέτετε προϊόντα μαναβικής;"
    }
])
# Η απάντηση του LLM
response.pretty_print()
````

## Παράδειγμα structured output

````python
# Ορίζουμε οτι πρέπει να απαντά πάντα σε json, και ένα παράδειγμα χρήσης.
# Αν υπάρχουν πολλαπλές μορφές απάντησης, πρέπει να ορίσουμε και τα αντίστοιχα παραδείγματα.
messages = [
    {
        "role": "system",
        "content": """
        Είσαι βοηθός eshop. Δουλειά σου είναι να βοηθάς πελάτες του eshop, προτείνοντάς τους προϊόντα και υπηρεσίες.

        Θα απαντάς πάντα σε JSON. Παράδειγμα:
        ```
        {
            "response": "Διαθέτουμε τα παρακάτω προϊόντα:",
            "products": [
                {
                    "title": "Προϊόν Α",
                    "price": 15.6,
                    "description": "Bla bla bla..."
                }
            ]
        }
        ```

        To eshop διαθέτει τα παρακάτω προϊόντα:

        - Ξυριστική μηχανή Α, τιμή 20.6, Περιγραφή: ασδφ
        - Ξυριστική μηχανή Β, τιμή 19.2, Περιγραφή: φδσα
        - Φακός, τιμή 10.0, Περιγραφή: Ένας φακός

        Αν δεν γνωρίζεις κάτι, μην αυτοσχεδιάσεις και πες οτι δεν το γνωρίζεις.
        """
    }
]
````

````python
response = model.invoke(input=[
    *messages,
    {
        "role": "user",
        "content": "Μπορείτε να μου προτείνετε ξυριστικές μηχανές;"
    }
])
response.pretty_print()
````
